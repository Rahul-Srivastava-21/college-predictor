```python
# === Stage1EDA-KCET : CELL 5 ===
# Temporal normalization + Tier threshold analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

NOTEBOOK_DIR = Path(".").resolve()
BASE = NOTEBOOK_DIR.parent
DATA_DIR = BASE / "data" / "processed_data"
STAGE1_DIR = DATA_DIR / "Stage1Files"
STAGE1_DIR.mkdir(exist_ok=True)

# Load engineering-only dataset
df = pd.read_csv(DATA_DIR / "KCET_split_engineering_only.csv", low_memory=False)

print("Rows:", len(df))

# 1. Year-wise max rank
year_max = df.groupby("Year")["Cutoff_Rank"].max().to_dict()
print("\nYear-wise max ranks:", year_max)

# 2. Normalized rank
df["Rank_norm"] = df.apply(lambda x: x["Cutoff_Rank"] / year_max[x["Year"]], axis=1)

df.to_csv(STAGE1_DIR / "kcet_normalized.csv", index=False)
print("Saved normalized dataset.")

# 3. Compare distributions
plt.figure(figsize=(12,4))
sns.boxplot(data=df, x="Year", y="Cutoff_Rank")
plt.title("Raw Cutoff Rank by Year (High Drift)")
plt.show()

plt.figure(figsize=(12,4))
sns.boxplot(data=df, x="Year", y="Rank_norm")
plt.title("Normalized Rank by Year (Stabilized)")
plt.show()

# 4. Tier thresholds (percentile-based)
college_medians = df.groupby("College_Code")["Cutoff_Rank"].median()

p20 = np.percentile(college_medians, 20)
p50 = np.percentile(college_medians, 50)
p80 = np.percentile(college_medians, 80)

print("\nSuggested Tier Cutoffs:")
print("Tier 1 upper cutoff (20th pct):", int(p20))
print("Tier 2 upper cutoff (50th pct):", int(p50))
print("Tier 3 lower cutoff (80th pct):", int(p80))

plt.figure(figsize=(6,10))
plt.boxplot(college_medians, vert=True)
plt.title("Distribution of College Median Cutoffs")
plt.ylabel("Median Cutoff Rank")
plt.show()
```

```python
# === Stage1EDA-KCET : CELL 6 ===
# Add TIER for each college and create Stage-2 ML-ready dataset

import pandas as pd
import numpy as np
from pathlib import Path

NOTEBOOK_DIR = Path(".").resolve()
BASE = NOTEBOOK_DIR.parent
DATA_DIR = BASE / "data" / "processed_data"
STAGE1_DIR = DATA_DIR / "Stage1Files"

df = pd.read_csv(STAGE1_DIR / "kcet_normalized.csv")

# Compute medians again (needed for mapping)
college_medians = df.groupby("College_Code")["Cutoff_Rank"].median()

# Tier thresholds
T1 = 56493
T2 = 87829

def assign_tier(median):
    if median <= T1:
        return 1
    elif median <= T2:
        return 2
    else:
        return 3

tier_map = college_medians.apply(assign_tier).to_dict()

df["College_Tier"] = df["College_Code"].map(tier_map)

print(df["College_Tier"].value_counts())

# Save Stage-2 input dataset
OUTPUT = DATA_DIR / "KCET_stage2_dataset.csv"
df.to_csv(OUTPUT, index=False)

print("\nStage-2 dataset saved →", OUTPUT)
```

```python
# ==========================================================
# FINAL SAVE-ALL CELL FOR MOVING INTO Stage2FE-KCET
# Saves every important variable, df, artifact, map, tier cutoffs
# into Version2/data/stage2_outputs
# ==========================================================

import os, json, pickle, joblib
from pathlib import Path
import pandas as pd

# -------------------------------
# Output directory
# -------------------------------
OUT_DIR = Path(
    r"D:\Courses\Global Academy of Technology\kcet-college-pred\Version2\data\stage2_outputs"
)
OUT_DIR.mkdir(parents=True, exist_ok=True)

print("Saving artifacts to:", OUT_DIR)

# -------------------------------
# Variables you want saved
# Add/remove names as needed
# -------------------------------
VARIABLES_TO_SAVE = {
    # DataFrames
    "df": "stage1_df.pkl",
    "df_clean": "df_clean.pkl",
    "df_normalized": "df_normalized.pkl",
    "stage2_df": "stage2_df.pkl",

    # Arrays / dicts
    "year_max": "year_max_ranks.json",
    "year_max_ranks": "year_max_ranks.json",
    "college_medians": "college_medians.pkl",
    "tier_map": "tier_map.json",
    "tier_cutoffs": "tier_cutoffs.json",

    # FE-related placeholders
    "unstable_branches": "unstable_branches.pkl",
    "feature_list": "feature_list.json",
    "college_tier_counts": "college_tier_counts.json",

    # Encoders/scalers (if exist)
    "scaler": "scaler.joblib",
    "rank_scaler": "rank_scaler.joblib",
    "label_encoders": "label_encoders.joblib",
    "target_encoders": "target_encoders.joblib",
}

saved_files = {}

# ==========================================================
# Helper functions
# ==========================================================

def save_json(obj, path):
    try:
        with open(path, "w", encoding="utf8") as f:
            json.dump(obj, f, indent=2, default=str)
        return True
    except:
        return False

def save_pickle(obj, path):
    try:
        with open(path, "wb") as f:
            pickle.dump(obj, f, protocol=pickle.HIGHEST_PROTOCOL)
        return True
    except:
        return False

def save_joblib(obj, path):
    try:
        joblib.dump(obj, path)
        return True
    except:
        return False

# ==========================================================
# MAIN SAVE LOOP
# ==========================================================
ns = globals()  # namespace

for var_name, filename in VARIABLES_TO_SAVE.items():
    if var_name not in ns:
        continue  # skip missing objects

    obj = ns[var_name]
    out_path = OUT_DIR / filename

    success = False

    # Decide format by extension
    if filename.endswith(".json"):
        success = save_json(obj, out_path)
    elif filename.endswith(".joblib"):
        success = save_joblib(obj, out_path)
    elif filename.endswith(".pkl"):
        success = save_pickle(obj, out_path)
    else:
        # fallback to pickle
        success = save_pickle(obj, out_path)

    if success:
        saved_files[var_name] = str(out_path)
    else:
        print(f"[WARN] Could not save {var_name}")

# ==========================================================
# Extra: Save Stage-2 CSV (if df exists)
# ==========================================================
if "stage2_df" in ns:
    try:
        csv_path = OUT_DIR / "stage2_dataset.csv"
        ns["stage2_df"].to_csv(csv_path, index=False)
        saved_files["stage2_df_csv"] = str(csv_path)
    except Exception as e:
        print("ERROR saving Stage2 CSV:", e)

# ==========================================================
# Manifest file
# ==========================================================
manifest = {
    "saved_files": saved_files,
    "total_saved": len(saved_files),
    "note": "List of all successfully saved artifacts for Stage2FE-KCET."
}

with open(OUT_DIR / "manifest.json", "w", encoding="utf8") as f:
    json.dump(manifest, f, indent=2)

print("\n===================================")
print(" STAGE 2 SAVE COMPLETE")
print(" Files written:", len(saved_files))
print(" Manifest:", OUT_DIR / "manifest.json")
print("===================================")
```
