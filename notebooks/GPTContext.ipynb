# COMPLETE DATASET DIAGNOSTIC FOR CHATGPT CONTEXT

This notebook provides exhaustive information about:
1. Raw dataset (df_optimized.csv) - BEFORE Stage 2
2. Processed datasets (Stage 2 outputs) - AFTER feature engineering
3. Column mappings, dtypes, and unique values
4. Label encoding mappings for reconstruction

**Goal**: Provide ChatGPT with exact structure to generate correct merging logic for Stage 3.

```python
# ============================================================================
# STEP 1: LOAD RAW DATASET (BEFORE STAGE 2)
# ============================================================================
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

print("=" * 100)
print("STEP 1: RAW DATASET ANALYSIS (df_optimized.csv - BEFORE STAGE 2)")
print("=" * 100)

# Load raw dataset
df_raw = pd.read_csv('kcet_ml_project/data/df_optimized.csv')

print(f"\n📊 SHAPE:")
print(f"   {df_raw.shape}")

print(f"\n📋 COLUMNS ({len(df_raw.columns)} total):")
for i, col in enumerate(df_raw.columns.tolist(), 1):
    print(f"   {i:2d}. {col}")

print(f"\n🔢 DATA TYPES:")
print(df_raw.dtypes.to_string())

print(f"\n📄 FIRST 10 ROWS (TRANSPOSED):")
print(df_raw.head(10).T.to_string())

print(f"\n" + "=" * 100)
```

```python
# ============================================================================
# STEP 1 CONTINUED: UNIQUE VALUES FOR KEY IDENTIFIER COLUMNS (RAW)
# ============================================================================

print("\n" + "=" * 100)
print("STEP 1 CONTINUED: UNIQUE VALUES IN RAW DATASET")
print("=" * 100)

# Check which identifier columns exist
identifier_cols = ['College_Code', 'College_Name', 'Branch', 'Category', 
                   'Category_Simplified', 'Exam_Type', 'Round', 'Year']

for col in identifier_cols:
    if col in df_raw.columns:
        print(f"\n📌 {col}:")
        print(f"   dtype: {df_raw[col].dtype}")
        print(f"   unique count: {df_raw[col].nunique()}")
        
        unique_vals = df_raw[col].unique()
        if len(unique_vals) <= 50:
            print(f"   unique values (all {len(unique_vals)}):")
            for val in sorted(unique_vals) if df_raw[col].dtype in ['int64', 'float64'] else unique_vals:
                print(f"      {val}")
        else:
            print(f"   unique values (first 30 of {len(unique_vals)}):")
            for val in list(unique_vals)[:30]:
                print(f"      {val}")
    else:
        print(f"\n❌ {col}: NOT FOUND in raw dataset")

print("\n" + "=" * 100)
```

```python
# ============================================================================
# STEP 2: LOAD STAGE 2 PROCESSED FILES (TRAIN)
# ============================================================================

print("\n" + "=" * 100)
print("STEP 2A: STAGE 2 PROCESSED - TRAIN SPLIT")
print("=" * 100)

train = pd.read_csv('kcet_ml_project/data/stage2_v2_corrected/train_stage2_final.csv')

print(f"\n📊 TRAIN SHAPE:")
print(f"   {train.shape}")

print(f"\n📋 TRAIN COLUMNS ({len(train.columns)} total):")
for i, col in enumerate(train.columns.tolist(), 1):
    print(f"   {i:2d}. {col}")

print(f"\n🔢 TRAIN DATA TYPES:")
print(train.dtypes.to_string())

print(f"\n📄 TRAIN FIRST 10 ROWS (TRANSPOSED):")
print(train.head(10).T.to_string())

print(f"\n🔍 IDENTIFIER COLUMNS CHECK (Train):")
identifier_cols = ["College_Code", "College_Name", "Branch", "Category", 
                   "Category_Simplified", "Exam_Type", "Round", "Year"]

for col in identifier_cols:
    if col in train.columns:
        print(f"   ✅ {col}: EXISTS")
        print(f"      dtype: {train[col].dtype}")
        print(f"      unique count: {train[col].nunique()}")
        unique_vals = train[col].unique()
        if len(unique_vals) <= 30:
            print(f"      unique values: {sorted(unique_vals) if train[col].dtype in ['int64', 'float64'] else list(unique_vals)}")
        else:
            print(f"      unique values (first 20): {list(unique_vals)[:20]}")
    else:
        print(f"   ❌ {col}: NOT FOUND")

print("\n" + "=" * 100)
```

```python
# ============================================================================
# STEP 2: LOAD STAGE 2 PROCESSED FILES (VAL)
# ============================================================================

print("\n" + "=" * 100)
print("STEP 2B: STAGE 2 PROCESSED - VAL SPLIT")
print("=" * 100)

val = pd.read_csv('kcet_ml_project/data/stage2_v2_corrected/val_stage2_final.csv')

print(f"\n📊 VAL SHAPE:")
print(f"   {val.shape}")

print(f"\n📋 VAL COLUMNS ({len(val.columns)} total):")
for i, col in enumerate(val.columns.tolist(), 1):
    print(f"   {i:2d}. {col}")

print(f"\n🔢 VAL DATA TYPES:")
print(val.dtypes.to_string())

print(f"\n📄 VAL FIRST 10 ROWS (TRANSPOSED):")
print(val.head(10).T.to_string())

print(f"\n🔍 IDENTIFIER COLUMNS CHECK (Val):")
identifier_cols = ["College_Code", "College_Name", "Branch", "Category", 
                   "Category_Simplified", "Exam_Type", "Round", "Year"]

for col in identifier_cols:
    if col in val.columns:
        print(f"   ✅ {col}: EXISTS")
        print(f"      dtype: {val[col].dtype}")
        print(f"      unique count: {val[col].nunique()}")
        unique_vals = val[col].unique()
        if len(unique_vals) <= 30:
            print(f"      unique values: {sorted(unique_vals) if val[col].dtype in ['int64', 'float64'] else list(unique_vals)}")
        else:
            print(f"      unique values (first 20): {list(unique_vals)[:20]}")
    else:
        print(f"   ❌ {col}: NOT FOUND")

print("\n" + "=" * 100)
```

```python
# ============================================================================
# STEP 2: LOAD STAGE 2 PROCESSED FILES (TEST)
# ============================================================================

print("\n" + "=" * 100)
print("STEP 2C: STAGE 2 PROCESSED - TEST SPLIT")
print("=" * 100)

test = pd.read_csv('kcet_ml_project/data/stage2_v2_corrected/test_stage2_final.csv')

print(f"\n📊 TEST SHAPE:")
print(f"   {test.shape}")

print(f"\n📋 TEST COLUMNS ({len(test.columns)} total):")
for i, col in enumerate(test.columns.tolist(), 1):
    print(f"   {i:2d}. {col}")

print(f"\n🔢 TEST DATA TYPES:")
print(test.dtypes.to_string())

print(f"\n📄 TEST FIRST 10 ROWS (TRANSPOSED):")
print(test.head(10).T.to_string())

print(f"\n🔍 IDENTIFIER COLUMNS CHECK (Test):")
identifier_cols = ["College_Code", "College_Name", "Branch", "Category", 
                   "Category_Simplified", "Exam_Type", "Round", "Year"]

for col in identifier_cols:
    if col in test.columns:
        print(f"   ✅ {col}: EXISTS")
        print(f"      dtype: {test[col].dtype}")
        print(f"      unique count: {test[col].nunique()}")
        unique_vals = test[col].unique()
        if len(unique_vals) <= 30:
            print(f"      unique values: {sorted(unique_vals) if test[col].dtype in ['int64', 'float64'] else list(unique_vals)}")
        else:
            print(f"      unique values (first 20): {list(unique_vals)[:20]}")
    else:
        print(f"   ❌ {col}: NOT FOUND")

print("\n" + "=" * 100)
```

```python
# ============================================================================
# STEP 3: LOAD X_TRAIN, X_VAL, X_TEST (FEATURES ONLY)
# ============================================================================

print("\n" + "=" * 100)
print("STEP 3: FEATURE MATRICES (X_train, X_val, X_test)")
print("=" * 100)

X_train = pd.read_csv('kcet_ml_project/data/stage2_v2_corrected/X_train_stage2.csv')
X_val = pd.read_csv('kcet_ml_project/data/stage2_v2_corrected/X_val_stage2.csv')
X_test = pd.read_csv('kcet_ml_project/data/stage2_v2_corrected/X_test_stage2.csv')

print(f"\n📊 SHAPES:")
print(f"   X_train: {X_train.shape}")
print(f"   X_val:   {X_val.shape}")
print(f"   X_test:  {X_test.shape}")

print(f"\n📋 FEATURE COLUMNS IN X_TRAIN ({len(X_train.columns)} total):")
for i, col in enumerate(sorted(X_train.columns.tolist()), 1):
    print(f"   {i:2d}. {col}")

print(f"\n🔍 IDENTIFIER COLUMNS CHECK IN X_TRAIN:")
identifier_cols = ["College_Code", "College_Name", "Branch", "Category", 
                   "Category_Simplified", "Exam_Type", "Round", "Year"]

identifiers_found = []
for col in identifier_cols:
    if col in X_train.columns:
        identifiers_found.append(col)
        print(f"   ✅ {col}: FOUND (dtype={X_train[col].dtype})")
    else:
        print(f"   ❌ {col}: NOT FOUND")

if identifiers_found:
    print(f"\n⚠️  WARNING: {len(identifiers_found)} identifier columns found in X_train!")
    print(f"   These should typically be removed for model training.")
else:
    print(f"\n✅ GOOD: No identifier columns in X_train (as expected)")

print("\n" + "=" * 100)
```

```python
# ============================================================================
# STEP 4: LABEL ENCODING MAPPINGS (RECONSTRUCT FROM RAW + PROCESSED)
# ============================================================================

print("\n" + "=" * 100)
print("STEP 4: LABEL ENCODING MAPPINGS (RAW vs PROCESSED)")
print("=" * 100)

print("\n🔍 EXAM_TYPE MAPPING:")
print("=" * 50)
if 'Exam_Type' in df_raw.columns and 'Exam_Type' in train.columns:
    print(f"RAW (df_raw):")
    print(f"   dtype: {df_raw['Exam_Type'].dtype}")
    print(f"   unique values: {sorted(df_raw['Exam_Type'].unique())}")
    
    print(f"\nPROCESSED (train):")
    print(f"   dtype: {train['Exam_Type'].dtype}")
    print(f"   unique values: {sorted(train['Exam_Type'].unique())}")
    
    # Try to reconstruct mapping
    if df_raw['Exam_Type'].dtype == 'object' and train['Exam_Type'].dtype in ['int64', 'float64']:
        print(f"\n📌 RECONSTRUCTED MAPPING:")
        # Get unique string values from raw
        raw_unique = sorted(df_raw['Exam_Type'].unique())
        processed_unique = sorted(train['Exam_Type'].unique())
        
        print(f"   Likely mapping (assuming alphabetical encoding):")
        for i, val in enumerate(raw_unique):
            print(f"      '{val}' → {i}")
        
        print(f"\n   Actual values in processed: {processed_unique}")
else:
    print("   ⚠️  Exam_Type not found in both datasets or already processed")

print("\n" + "=" * 50)
print("\n🔍 CATEGORY MAPPING:")
print("=" * 50)
if 'Category' in df_raw.columns:
    print(f"RAW (df_raw):")
    print(f"   dtype: {df_raw['Category'].dtype}")
    raw_category_unique = df_raw['Category'].unique()
    if len(raw_category_unique) <= 50:
        print(f"   unique values ({len(raw_category_unique)} total):")
        for val in sorted(raw_category_unique):
            print(f"      {val}")
    else:
        print(f"   unique values (first 30 of {len(raw_category_unique)}):")
        for val in list(raw_category_unique)[:30]:
            print(f"      {val}")
    
    if 'Category' in train.columns:
        print(f"\nPROCESSED (train) - Category column EXISTS:")
        print(f"   dtype: {train['Category'].dtype}")
        print(f"   unique values: {sorted(train['Category'].unique())}")
    else:
        print(f"\nPROCESSED (train) - Category column REMOVED")
else:
    print("   ⚠️  Category not found in raw dataset")

print("\n" + "=" * 50)
print("\n🔍 CATEGORY_SIMPLIFIED MAPPING:")
print("=" * 50)
if 'Category_Simplified' in df_raw.columns:
    print(f"RAW (df_raw):")
    print(f"   dtype: {df_raw['Category_Simplified'].dtype}")
    print(f"   unique values: {sorted(df_raw['Category_Simplified'].unique())}")
    
    if 'Category_Simplified' in train.columns:
        print(f"\nPROCESSED (train):")
        print(f"   dtype: {train['Category_Simplified'].dtype}")
        print(f"   unique values: {sorted(train['Category_Simplified'].unique())}")
    else:
        print(f"\nPROCESSED (train) - Category_Simplified column REMOVED")
else:
    print("   ⚠️  Category_Simplified not found in raw dataset")

print("\n" + "=" * 100)
```

```python
# ============================================================================
# STEP 5: SIDE-BY-SIDE COMPARISON (10 RANDOM ROWS)
# ============================================================================

print("\n" + "=" * 100)
print("STEP 5: SIDE-BY-SIDE COMPARISON (RAW vs PROCESSED)")
print("=" * 100)

# Sample 10 random rows (use same seed for reproducibility)
np.random.seed(42)
sample_indices = np.random.choice(len(df_raw), size=min(10, len(df_raw)), replace=False)

print("\n🔍 10 RANDOM ROWS FROM RAW DATASET:")
print("=" * 50)
raw_key_cols = ['College_Code', 'Branch', 'Category', 'Exam_Type', 'Year', 'Round']
existing_raw_cols = [col for col in raw_key_cols if col in df_raw.columns]

df_raw_sample = df_raw.iloc[sample_indices][existing_raw_cols]
print(df_raw_sample.to_string())

print("\n" + "=" * 50)
print("\n🔍 10 RANDOM ROWS FROM PROCESSED TRAIN DATASET:")
print("=" * 50)
train_sample = train.sample(10, random_state=42)

# Show Year, Round, Exam_Type, and first 15 numeric columns
processed_key_cols = ['Year', 'Round', 'Exam_Type']
existing_processed_cols = [col for col in processed_key_cols if col in train.columns]

# Add first 10 numeric feature columns
numeric_cols = [col for col in train.columns if col not in ['Cutoff_Rank'] and train[col].dtype in ['int64', 'float64']]
display_cols = existing_processed_cols + numeric_cols[:10]

print(f"Displaying columns: {display_cols}")
print(train_sample[display_cols].to_string())

print("\n" + "=" * 100)
```

```python
# ============================================================================
# STEP 6: SUMMARY FOR CHATGPT - KEY TAKEAWAYS
# ============================================================================

print("\n" + "=" * 100)
print("STEP 6: SUMMARY FOR CHATGPT - KEY INFORMATION")
print("=" * 100)

# Check if all variables are loaded (in case cells were run out of order)
# Note: val might be a string if cell 6 didn't run properly - reload it
try:
    if not hasattr(val, 'shape'):
        print("\n⚠️  Reloading val dataset (was not a DataFrame)...")
        val = pd.read_csv('kcet_ml_project/data/stage2_v2_corrected/val_stage2_final.csv')
        print("   ✅ Val dataset reloaded")
except:
    pass

# Now proceed with summary
try:
    print("\n📌 DATASET SHAPES:")
    print(f"   Raw (df_optimized.csv):           {df_raw.shape}")
    print(f"   Processed Train:                  {train.shape}")
    print(f"   Processed Val:                    {val.shape}")
    print(f"   Processed Test:                   {test.shape}")
    print(f"   X_train (features only):          {X_train.shape}")

    print("\n📌 COLUMNS IN RAW vs PROCESSED:")
    raw_cols = set(df_raw.columns)
    train_cols = set(train.columns)
    removed_cols = raw_cols - train_cols
    added_cols = train_cols - raw_cols
    common_cols = raw_cols & train_cols

    print(f"\n   Common columns: {len(common_cols)}")
    print(f"   Removed in Stage 2: {len(removed_cols)}")
    if len(removed_cols) <= 20:
        print(f"      {sorted(removed_cols)}")
    else:
        print(f"      First 20: {sorted(list(removed_cols))[:20]}")
        
    print(f"   Added in Stage 2: {len(added_cols)}")
    if len(added_cols) <= 20:
        print(f"      {sorted(added_cols)}")
    else:
        print(f"      First 20: {sorted(list(added_cols))[:20]}")

    print("\n📌 KEY IDENTIFIERS THAT CAN BE USED FOR MERGING:")
    merge_keys = []
    for col in ['Year', 'Round', 'Exam_Type']:
        if col in df_raw.columns and col in train.columns:
            merge_keys.append(col)
            print(f"   ✅ {col}")
            print(f"      RAW dtype:       {df_raw[col].dtype}")
            print(f"      PROCESSED dtype: {train[col].dtype}")
            if df_raw[col].dtype != train[col].dtype:
                print(f"      ⚠️  DTYPE MISMATCH - conversion needed!")

    if not merge_keys:
        print("   ❌ NO COMMON MERGE KEYS FOUND")
    else:
        print(f"\n   Suggested merge strategy:")
        print(f"      Use columns: {merge_keys}")
        print(f"      Handle dtype conversions if needed")

    print("\n📌 MISSING IDENTIFIERS IN PROCESSED DATA:")
    missing_identifiers = ['College_Code', 'College_Name', 'Branch', 'Category']
    for col in missing_identifiers:
        if col in df_raw.columns and col not in train.columns:
            print(f"   ⚠️  {col}: In RAW but REMOVED in Stage 2")
            print(f"      Can use target encodings or encoded features as proxy")

    print("\n📌 LABEL ENCODING INFO:")
    if 'Exam_Type' in df_raw.columns and 'Exam_Type' in train.columns:
        if df_raw['Exam_Type'].dtype == 'object' and train['Exam_Type'].dtype in ['int64', 'float64']:
            print(f"   Exam_Type: Label encoded from string → int")
            print(f"      Use LabelEncoder or manual mapping to reconstruct")

    print("\n" + "=" * 100)
    print("✅ DIAGNOSTIC COMPLETE - All information printed above")
    print("=" * 100)
    
except Exception as e:
    print(f"\n❌ ERROR generating summary: {e}")
    print(f"   Please ensure all previous cells ran successfully.")
```
